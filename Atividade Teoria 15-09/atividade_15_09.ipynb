### Questão 1

```python
import numpy as np
import control as cnt
import matplotlib.pyplot as plt
import plotly.graph_objects as go
#considerando uma função de transferencia em malha aberta FT=k/(tau*s+1)
k=60
tau=40.5
Theta = 2.5 # atraso de propagação
```

```python
#parâmetros do controlador kp+kp/(Ti*s)+kp*Td*s

# ZIEGLER NICHOLS
kp=0.54
Ti=3
Td=7.5

num = np. array ([k])
den = np. array ([tau , 1])
H = cnt.tf(num , den)
n_pade = 10
( num_pade , den_pade ) = cnt.pade ( Theta , n_pade )
H_pade = cnt.tf( num_pade , den_pade )
numkp = np. array ([kp])
denkp = np. array ([1])
numki = np. array ([kp])
denki = np. array ([Ti,0])
numkd = np. array ([kp*Td,0])
denkd = np. array ([1])
Hkp = cnt.tf(numkp , denkp)
Hki=cnt.tf(numki , denki)
Hkd=cnt.tf(numkd , denkd)
Hctrl1 = cnt.parallel (Hkp , Hki)
Hctrl = cnt.parallel (Hctrl1 , Hkd)
Hs = cnt.series (H , H_pade)
Hdel = cnt.series (Hs , Hctrl)
Hcl = cnt.feedback(Hdel, 1)


t = np . linspace (0 , 40 , 100)
(t , y ) = cnt.step_response ( Hcl, t )
plt.plot (t , y )
plt.xlabel ( ' t [ s ] ')
plt.ylabel('Amplitude')
plt.title('Controle proporcional-integral-derivativo')

plt.grid ()
plt.show()
```

O método de Ziegler–Nichols tende a produzir ajustes mais agressivos e respostas rápidas, porém com maior possibilidade de sobrevalor e oscilações. O método CHR busca uma resposta mais controlada, podendo ser configurado para eliminar o sobrevalor. Já o método IMC permite ajustar o compromisso entre rapidez e robustez por meio de seu parâmetro de projeto, sendo geralmente mais adequado quando se deseja maior robustez a incertezas e atrasos da planta.
